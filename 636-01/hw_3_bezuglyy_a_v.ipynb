Climbing Stairs

```python
def climbing_stairs_memo(n):
    memo = {}

    def dp(i):
        if i <= 1:
            return 1

        if i not in memo:
            memo[i] = dp(i - 1) + dp(i - 2)

        return memo[i]

    return dp(n)
```

```python
def climbing_stairs_tabulaion(n):
    if n <= 1:
        return 1

    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
```

```python
def climbing_stairs_optimized_tabulation(n):
    if n <= 1:
        return 1

    prev1 = 1
    prev2 = 1

    for i in range(2, n + 1):
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current

    return prev1
```

```python
print("n = 2:", climbing_stairs_memo(2))
print("n = 3:", climbing_stairs_memo(3))

print("Табуляция:", climbing_stairs_tabulaion(5))
print("Оптимизированный вариант:", climbing_stairs_optimized_tabulation(5))
```

House Robber

```python
def house_robber_memo(nums):
    memo = {}

    def dp(i):
        if i < 0:
            return 0

        if i not in memo:
            memo[i] = max(
                dp(i - 1),
                nums[i] + dp(i - 2)
            )

        return memo[i]

    return dp(len(nums) - 1)
```

```python
def house_robber_tabulation(nums):
    if not nums:
        return 0

    if len(nums) == 1:
        return nums[0]

    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])

    for i in range(2, len(nums)):
        dp[i] = max(dp[i - 1], nums[i] + dp[i - 2])

    return dp[-1]
```

```python
def house_robber_optimized_tabulation(nums):
    prev2 = 0
    prev1 = 0

    for num in nums:
        current = max(prev1, prev2 + num)
        prev2 = prev1
        prev1 = current

    return prev1
```

```python
nums1 = [1, 2, 3, 1]
nums2 = [2, 7, 9, 3, 1]
nums3 = []
nums4 = [5]
nums5 = [2, 7]

print(house_robber_memo(nums1))
print(house_robber_tabulation(nums2))
print(house_robber_optimized_tabulation(nums3))
print(house_robber_optimized_tabulation(nums4))
print(house_robber_optimized_tabulation(nums5))
```

```python
def house_robber_circular(nums):
    if len(nums) <= 1:
        return nums[0] if nums else 0

    return max(
        house_robber_optimized_tabulation(nums[:-1]),
        house_robber_optimized_tabulation(nums[1:])
    )

print(house_robber_circular([2, 3, 2]))
print(house_robber_circular([1, 2, 3, 1]))
```
